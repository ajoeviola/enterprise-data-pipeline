```python
# Import Spark
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("DataIngestion").getOrCreate()

# Read Sales Data
df = spark.table("workspace.default.customer_data")

# Show data
df.display()
```

```python
# Create or overwrite a managed Delta table
df.write.format("delta").mode("overwrite").saveAsTable("cleaned_customer_data")
```
